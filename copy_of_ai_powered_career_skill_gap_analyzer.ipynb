```python
def generate_report(career, skills, education, experience):

    prompt = f"""
You are an expert AI Career Counselor.

A candidate has provided the following information.

Target Career:
{career}

Education:
{education}

Experience Level:
{experience}

Current Skills:
{skills}

Perform the following tasks.

1. Identify the top 10 important skills required for the target career.

2. Compare the candidate's skills with the required skills.

3. Calculate an estimated Career Match Score (0-100).

4. Display

• Available Skills

• Missing Skills

5. Explain every missing skill.

Include:
• Why it is important
• Industry usage
• Difficulty level

6. Create a Personalized 3-Month Learning Roadmap.

Month 1

Month 2

Month 3

7. Recommend five Certifications.

8. Recommend five Resume Projects.

9. Mention Career Opportunities after learning the missing skills.

10. Give Interview Preparation Tips.

11. Give Final Career Advice.

Format everything professionally using Markdown headings and bullet points.
"""

    try:

        response = client.models.generate_content(
            model="models/gemini-3.6-flash",
            contents=prompt
        )

        return response.text

    except Exception as e:

        return f"Error: {str(e)}"
```

```python
with gr.Blocks(theme=gr.themes.Soft()) as app:

    gr.Markdown(
        """
# 🎯 AI-Powered Universal Career Skill Gap Analyzer

"""
    )

    with gr.Row():

        with gr.Column():

            career = gr.Textbox(
                label="🎯 Target Career",
                placeholder="Example: AI Engineer"
            )

            skills = gr.Textbox(
                label="💻 Current Skills",
                lines=4,
                placeholder="Example: Python, SQL, Communication"
            )

            education = gr.Textbox(
                label="🎓 Education"
            )

            experience = gr.Dropdown(

                choices=[
                    "Beginner",
                    "Intermediate",
                    "Advanced"
                ],

                value="Beginner",

                label="Experience Level"

            )

            analyze = gr.Button(
                "🚀 Analyze Career",
                variant="primary"
            )

        with gr.Column():

            output = gr.Markdown()

    analyze.click(

        fn=generate_report,

        inputs=[
            career,
            skills,
            education,
            experience
        ],

        outputs=output

    )
```
