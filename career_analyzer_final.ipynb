```python
def generate_report(career, skills, education, experience):

    prompt = f"""
You are an Expert AI Career Counselor.

Analyze the following candidate.

Candidate Details

Target Career:
{career}

Current Skills:
{skills}

Education:
{education}

Experience Level:
{experience}

Instructions:

The user may enter ANY career.

Examples include:

Doctor
Lawyer
Pilot
Teacher
Chef
Architect
Fashion Designer
Scientist
Mechanical Engineer
Data Scientist
Cyber Security Analyst
Army Officer
Professor
Business Analyst
Marine Biologist

Do NOT assume the career belongs to IT.

First understand the profession.

Then generate a professional report.

The report must include the following sections.

# Career Overview

Briefly explain the selected career.

# Required Skills

List the top 10 skills required.

# Current Skills Analysis

Identify which of the user's skills match the required skills.

# Missing Skills

Identify the missing skills.

Explain why each missing skill is important.

# Career Match Score

Estimate a score between 0 and 100.

Explain the score.

# Three Month Learning Roadmap

Month 1

Month 2

Month 3

# Recommended Certifications

Recommend five certifications.

# Recommended Projects or Practical Activities

Recommend five suitable projects or practical learning activities.

# Interview Preparation Tips

Provide interview preparation guidance.

# Career Opportunities

Mention possible job roles.

# Final Career Advice

Provide professional career guidance.

If the entered career is invalid or cannot be recognized, politely ask the user to enter a valid career.

Return the report using proper Markdown headings and bullet points.
"""

    try:

        response = client.chat.completions.create(

            model="gpt-4.1-nano",

            messages=[

                {
                    "role":"system",
                    "content":"You are an experienced AI Career Guidance Expert."
                },

                {
                    "role":"user",
                    "content":prompt
                }

            ]

        )

        return response.choices[0].message.content

    except Exception as e:

        return f"Error : {e}"
```

```python
with gr.Blocks(theme=gr.themes.Soft()) as app:

    gr.Markdown(
    """
# 🎯 AI-Powered Universal Career Skill Gap Analyzer


"""
    )

    with gr.Row():

        with gr.Column():

            career = gr.Textbox(

                label="Target Career",

                placeholder="Example : Doctor"

            )

            skills = gr.Textbox(

                label="Current Skills",

                lines=4,

                placeholder="Example : Communication, Patient Care"

            )

            education = gr.Textbox(

                label="Education",

                placeholder="Example : MBBS"

            )

            experience = gr.Dropdown(

                choices=[
                    "Beginner",
                    "Intermediate",
                    "Advanced"
                ],

                value="Beginner",

                label="Experience Level"

            )

            analyze = gr.Button(

                "Analyze Career",

                variant="primary"

            )

        with gr.Column():

            output = gr.Markdown()

    analyze.click(

        fn=generate_report,

        inputs=[

            career,

            skills,

            education,

            experience

        ],

        outputs=output

    )
```

```python
app.launch(debug=True)
```
